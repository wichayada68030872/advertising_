```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r'C:\Users\boomt\Downloads\Advertising.csv')
df = df.drop(columns='Unnamed: 0')
df.head()
```

```python
names = {'TV': 'โฆษณาทางทีวี (TV)', 'Radio': 'โฆษณาทางวิทยุ (Radio)',
         'Newspaper': 'โฆษณาทางหนังสือพิมพ์ (Newspaper)', 'Sales': 'ยอดขาย (Sales)'}

fig, axes = plt.subplots(2, 2, figsize=(12, 9))
for ax, col in zip(axes.flatten(), df.columns):
    ax.hist(df[col], bins=15, color='steelblue', edgecolor='white')
    ax.set_title( names[col])
    ax.set_xlabel(col)
    ax.set_ylabel('ความถี่')
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 4, figsize=(14, 5))
for ax, col in zip(axes, df.columns):
    ax.boxplot(df[col])
    ax.set_title(col)
    ax.set_xticks([])
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for ax, col in zip(axes, ['TV', 'Radio', 'Newspaper']):
    ax.scatter(df[col], df['Sales'], color='steelblue', alpha=0.7)
    r = df[col].corr(df['Sales'])
    ax.set_title(col + ' กับ Sales ')
    ax.set_xlabel('งบค่าโฆษณา ' + col)
    ax.set_ylabel('Sales')
plt.tight_layout()
plt.show()
```
